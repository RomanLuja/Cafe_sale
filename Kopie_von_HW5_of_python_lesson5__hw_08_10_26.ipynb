10. Створи три змінні на основі `students`:
- `ref` — просте присвоєння  (b = a);
- `shallow` — поверхнева копія (b = copy(a));
- `deep` — глибока копія (b = copy.deepcopy(a)).

Потім у `shallow` додай студентці `"Sofia"` оцінку 100, а з `ref` видали останнього студента.

Виведи кількість студентів та оцінки студентки `"Sofia"` в `students`, `ref`, `shallow` і `deep`. Поясни в коментарі отримані результати

````python
import copy

ref = students                    # просте присвоєння: друге ім'я для того самого списку
shallow = copy.copy(students)     # поверхнева копія: новий список, але ті самі словники всередині
deep = copy.deepcopy(students)    # глибока копія: новий список і нові словники

ref = students                    # просте присвоєння: друге ім'я для того самого списку
shallow = copy.copy(students)     # поверхнева копія: новий список, але ті самі словники всередині
deep = copy.deepcopy(students)    # глибока копія: новий список і нові словники

# Додаємо студентці Sofia оцінку 100 через shallow
for s in shallow:
    if s['name'] == 'Sofia':
        s['scores'].append(100)

# Видаляємо останнього студента через ref
ref.pop()

# Оцінки Sofia в заданому списку
def sofia_scores(lst):
    return next(s['scores'] for s in lst if s['name'] == 'Sofia')

for name, lst in [('students', students), ('ref', ref),
                  ('shallow', shallow), ('deep', deep)]:
    print(name, len(lst), sofia_scores(lst))

# Пояснення результатів:
# 1) ref — це те саме посилання, що й students (один об'єкт у пам'яті),
#    тому ref.pop() видалив останнього студента і зі students, і з ref:
#    у них однакова кількість студентів (на одного менше, ніж було).
# 2) shallow — новий зовнішній список, тому pop() через ref його не зачепив:
#    у shallow кількість студентів лишилась початковою.
#    Але словники всередині shallow — ті самі об'єкти, що й у students,
#    тому оцінка 100, додана через shallow, з'явилась у students і ref.
# 3) deep — повністю незалежна копія (і список, і вкладені словники):
#    у ній не змінилась ні кількість студентів, ні оцінки Sofia.
````

- витрати на рекламу за кожен місяць року по порядку;
- розмір екрана: ширина та висота в пікселях;

12. Код нижче має:
- вивести кількість відвідувачів у четвер, а якщо даних немає — `0`;
- вивести найбільшу кількість відвідувачів за день;
- перевірити, чи був день, коли прийшло 95 відвідувачів.

Знайди й виправ помилки, поясни кожну в коментарі

````python
visits = {"Mon": 120, "Tue": 95, "Wed": 140}

# Помилка 1: visits["Thu"] викликає KeyError, бо ключа "Thu" у словнику немає.
# Виправлення: .get("Thu", 0) повертає 0, якщо ключа немає.
print(visits.get("Thu", 0))        # 0

# Помилка 2: max(visits) шукає максимум серед ключів (назв днів),
# тому поверне "Wed" (за алфавітом), а не кількість відвідувачів.
# Виправлення: брати максимум серед значень через .values().
print(max(visits.values()))        # 140

# Помилка 3: 95 in visits перевіряє лише ключі (назви днів),
# тому завжди дасть False, бо 95 не є назвою дня.
# Виправлення: шукати серед значень через .values().
print(95 in visits.values())       # True
````

13. Код нижче має:
- створити множину категорій і додати до неї категорію `"food"`;
- створити кортеж з одного дня `"Mon"` і отримати новий кортеж з двох днів `("Mon", "Tue")`;
- створити список покупок і додати до нього `"tea"`, щоб вийшов список з трьох елементів.

Знайди й виправ помилки, поясни кожну в коментарі

````python
# Блок 1: множина категорій
# Помилка: categories = {} створює порожній СЛОВНИК (dict), а не множину,
# тому в нього немає методу .add() і виникає AttributeError.
# Виправлення: порожню множину створюємо через set().
categories = set()
categories.add("food")
print(categories)                  # {'food'}

# Блок 2: кортеж з днів
# Виправлення: кортеж з одного елемента пишемо з комою: ("Mon",).
# Тоді + з кортежем ("Tue",) дає новий кортеж з двох днів.
week = ("Mon",)
week = week + ("Tue",)
print(week)                        # ('Mon', 'Tue')

# Блок 3: список покупок
# Помилка: append() змінює список на місці і повертає None,
# тому shopping = shopping.append("tea") перезаписує список значенням None.
# Виправлення: викликаємо append() без присвоєння.
shopping = ["milk", "bread"]
shopping.append("tea")
print(shopping)                    # ['milk', 'bread', 'tea']
````

15. Збери звіт у словник `report` з ключами:
- `"titles"` — скільки різних назв книг є в бібліотеці;
- `"copies_left"` — скільки примірників залишилося загалом;
- `"out_of_stock"` — чи є книга, якої не залишилося в наявності (`True` / `False`);
- `"unique_readers"` — скільки різних читачів приходило за два дні;
- `"loyal_readers"` — множина читачів, які приходили обидва дні.

Виведи `report`

````python
report = {
    "titles": len(books),
    "copies_left": sum(books.values()),
    "out_of_stock": 0 in books.values(),
    "unique_readers": len(readers_mon | readers_tue),
    "loyal_readers": readers_mon & readers_tue,
}

print(report)
````
